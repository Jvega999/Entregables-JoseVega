### Identificar variables categoricas que presenten inconsistencias y estandarizarlas

```python
df["Animal type"] = (df["Animal type"] # Unificamos el dato Lynx y Lynx?
                    .str.replace("?" , "" , regex = False)
                    )

df["Animal type"] = (df["Animal type"] # Unificamos los datos European bison
                    .str.replace("™" , "" , regex = False)
                    .str.replace("bisson" , "bison" , regex = False)
                    .str.replace("European buster" , "European bison" , regex = False)
                    )

df["Animal type"] = (df["Animal type"] # Unificamos los datos red squirrel
                     .str.replace("red squirrell" , "red squirrel" , regex = False)
                     .str.replace("red squirel" , "red squirrel" , regex = False)
                     )

df["Animal type"] = (df["Animal type"] # Unificamos los datos hedgehod
                     .str.replace("ledgehod" , "hedgehog" , regex = False)
                     .str.replace("Hedgehog" , "hedgehog" , regex = False)
                     .str.replace("wedgehod" , "hedgehog" , regex = False)

                     )

df["Animal type"].value_counts()
```

```python
# Rellenar los valores nulos de 'Weight kg' usando el promedio correspondiente a cada especie ('Animal type')
df["Weight kg"] = df["Weight kg"].fillna(df.groupby("Animal type")["Weight kg"].transform("mean"))

# Hacemos lo mismo para la columna de longitud corporal 'Body Length cm'
df["Body Length cm"] = df["Body Length cm"].fillna(df.groupby("Animal type")["Body Length cm"].transform("mean"))

# Verificamos si aún quedan valores nulos en estas columnas de interés
print("Valores nulos restantes:")
display(df[["Weight kg", "Body Length cm"]].isnull().sum())
```
